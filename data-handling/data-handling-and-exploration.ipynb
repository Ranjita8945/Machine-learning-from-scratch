```python
df['Age'].notnull()
    
```

```python
conditions = [
    df['Age'] < 18,
    df['Age'] >= 18
]

choices = ['Child', 'Adult']

df['Age_Group'] = np.select(
    conditions,
    choices,
    default='Unknown'
)

df['Age_Group'].value_counts()    
```

```python
df.groupby('Sex')['Survived'].mean()
```

```python
df.groupby(['Sex','Pclass'])['Survived'].mean()
```

```python
df['Family_Size'] = df['SibSp'] + df['Parch'] + 1
```

```python
df[['Family_Size','SibSp','Parch']].head(10)
```

```python
df.groupby(['Family_Size'])['Survived'].mean() * 100
```

```python
df.groupby(['Family_Size'])['Survived'].agg(['count',
    'mean'])
```

```python
conditions = [
    df['Family_Size'] == 1,
    df['Family_Size']  <= 4,
]

choices = ['Alone', 'Small Family']

df['family_Group'] = np.select(
    conditions,
    choices,
    default='Large family'
)

df['family_Group'].value_counts()   
```

```python
df.groupby(['family_Group'])['Survived'].agg(['count',
    'mean']) * 100
```

```python
family_survival = df.groupby('family_Group')['Survived'].mean() * 100
```

```python
import matplotlib.pyplot as plt
family_survival.plot(kind='bar')
plt.title('Survival Rate by Family Category')
plt.xlabel('Family Category')
plt.ylabel('Survival Rate (%)')
for i, value in enumerate(family_survival):
    plt.text(i, value, f'{value:.1f}%', ha='center')
plt.show()
```

```python
Q1 = df['Age'].quantile(0.25)
Q2 =  df['Age'].quantile(0.75)

IQR = Q2 - Q1

print("Q1", Q1)
print("Q2", Q2)
print("IQR", IQR)
```

```python
lb = Q1 - 1.5 * IQR
ub = Q2 + 1.5 * IQR

print(lb)
print(ub)
```

```python
outliers = df[(df['Age'] < lb) | (df['Age'] > ub)]
print(outliers.shape[0])
```

# Feature engineering

### Feature engineering means creating new, meaningful columns from existing data to uncover patterns that weren't immediately visible.

```python
df['Name'].head(10)
```

```python
df['Title'] = df['Name'].str.extract(r',\s*([^.]*)\.')[0]
```

```python
df['Title'].value_counts()
```

```python
df['Title'].nunique()
```

```python
title_counts = df['Title'].value_counts()
rare_titles = title_counts[title_counts < 10]
print(rare_titles)
```

```python
df['Title_group'] = np.where(
    df['Title'].isin(['Mr', 'Mrs', 'Miss', 'Master']),
    df['Title'],
    'Rare'
)
```

```python
df['Title_group'].value_counts()
```

```python
df.groupby('Title_group')['Survived'].agg(['count', 'mean']).sort_values(
    by='mean', ascending=False
)
```
