```python
conditions = [
    df['Age'] < 18,
    df['Age'] >= 18
]

choices = ['Child', 'Adult']

df['Age_Group'] = np.select(
    conditions,
    choices,
    default='Unknown'
)

df['Age_Group'].value_counts()    
```
